## 5. Desenvolvimento Dashboard de visualização

No desenvolvimento da dashboard de visualização de rotas realizadas por três robots underwater, o repositório foi dividido de maneira a separar o desenvolvimento da interface do utilizador (layout) e da lógica de funcionamento (callbacks), de maneira a facilitar tarefas de manutenção e leitura do código. 
Assim, existem os seguintes ficheiros python: config.py onde se encontram centralizadas todas as constantes, leituras do ficheiro de configuração (config.yaml), caminhos para a pasta de logs fornecidos, as cores associadas a cada robô, e ainda as funções auxiliares de mapeamento e identificação de referências.
O layout.py é responsável por definir a estrutura visual da página, nomeadamente o cabeçalho, as secções divididas que organizam o mapa 2D, a vista 3D, o radar, os gráficos temporais daprofundidade e velocidade e a barra inferior de reprodução, onde é possível aumentar e diminuir o ritmo de reprodução. O callbacks.py contém todos os callbacks do Dash que gerem o dinamismo da dashboard. É aqui que o movimento dos robôs é sincronizado com a barra de reprodução, calculadas as trajetórias, atualizados os gráficos do Plotly e gerada a perspetiva 3D.

### 5.1. Transformações de coordenadas e matrizes de rotação

O dashboard usa duas matrizes de rotação, ambas implementadas com NumPy no ficheiro `callbacks.py`.

**Orientação do robô** Cada registo dos logs inclui a orientação do robô como quaternião unitário $q = (q_x, q_y, q_z, q_w)$. Para desenhar os eixos do robô na bússola 3D, o quaternião é convertido na matriz de rotação

$$
R(q)=\begin{bmatrix}
1-2(q_y^2+q_z^2) & 2(q_xq_y-q_zq_w) & 2(q_xq_z+q_yq_w)\\
2(q_xq_y+q_zq_w) & 1-2(q_x^2+q_z^2) & 2(q_yq_z-q_xq_w)\\
2(q_xq_z-q_yq_w) & 2(q_yq_z+q_xq_w) & 1-2(q_x^2+q_y^2)
\end{bmatrix}
$$

As colunas de $R$ são os eixos $X$, $Y$ e $Z$ do robô expressos no referencial do mundo, e são esses vetores que a bússola desenha.

**Posição relativa entre robôs** As posições dos logs estão em coordenadas geodésicas $(\varphi, \lambda, h)$. Para calcular distâncias entre robôs, cada posição é convertida para ECEF

$$
\begin{aligned}
X &= (N+h)\cos\varphi\cos\lambda, \quad Y = (N+h)\cos\varphi\sin\lambda, \quad Z = \big(N(1-e^2)+h\big)\sin\varphi,\\
N &= \frac{a}{\sqrt{1-e^2\sin^2\varphi}},
\end{aligned}
$$

e a diferença para a origem $\Delta = \mathbf{p}_{ECEF}-\mathbf{p}_{ECEF,0}$ é rodada para o referencial local NED, centrado em $(\varphi_0,\lambda_0,h_0)$:

$$
\begin{bmatrix}N\\E\\D\end{bmatrix}=
\begin{bmatrix}
-\sin\varphi_0\cos\lambda_0 & -\sin\varphi_0\sin\lambda_0 & \cos\varphi_0\\
-\sin\lambda_0 & \cos\lambda_0 & 0\\
-\cos\varphi_0\cos\lambda_0 & -\cos\varphi_0\sin\lambda_0 & -\sin\varphi_0
\end{bmatrix}\Delta
$$

Esta matriz é ortogonal e resulta da composição de duas rotações elementares: uma em torno de $Z$ pela longitude $\lambda_0$ e outra em torno de $Y$ pelo ângulo $-(\varphi_0+90^\circ)$. A partir de $(N,E,D)$ calculam-se a distância horizontal $\sqrt{N^2+E^2}$, a distância 3D $\sqrt{N^2+E^2+D^2}$ e o azimute $\operatorname{atan2}(E,N)$, medido a partir do norte no sentido horário, como no sistema AER apresentado nas aulas.

## 7. Conclusões e Trabalho Futuro

O desenvolvimento desta dashboard provou ser uma solução para a visualização e análise integrada de logs de navegação de robôs autónomos em ambiente subaquático. Este projeto permitiu ainda abrir horizontes nomeadamente nas capacidades da linguagem python e no desenvolvimento que as suas bibliotecas proporcionam. Sem dúvida que esta foi uma vantagem para o grupo, apesar da dificuldade sentida no início do desenvolvimento do projeto, dada a reduzida experiência com Python, o que levou a um período de ambientação que, em conjunto com ferramentas de IA, conseguimos ir alcançando os objetivos propostos, ganhando desta forma, mais autonomia ao longo do desenvolvimento. Para além disso, dado os objetivos propostos inicialmente, tivemos que transportar conhecimentos teóricos para o desenvolvimento desta dashboard, nomeadamente rotações de matrizes, transformações de coordenadas.

A integração do mapa geográfico em 2D com suporte OpenSeaMap, de uma vista tridimensional com cota de profundidade e nível do mar, para além de um radar de proximidade baseado em conversão de coordenadas geodésicas para NED e de gráficos temporais sincronizados permitiu ao grupo perceber e monitorizar em detalhe o comportamento e a trajetória destes veículos, pelo que, em caso de necessidade de avaliar uma determinada missão, a utilização desta dashboard passa por uma resposta a essa mesma necessidade. 

Como perspetivas de evolução e melhoria contínua para trabalho futuro, consideramos importante a implementação de um alerta estilo pop-up, caso os veículos possam colidir ou se encontrem a uma distância de risco. Outra oportunidade de melhoria seria dotar a dashboard com capacidade para operar em tempo real, ou seja, através de uma ligação WebSocket, permitir que sejam rececionadas mensagens de maneira a atualizar a dashboard, permitindo assim utilizar a mesma não só com ficheiros CSV com campos preenchidos, como foi o caso.
